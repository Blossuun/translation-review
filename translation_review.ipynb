## 표시 확인 (가상 예시)

하이라이트 색·번호, 좌우 2단, `<details>` 접기가 보이는지, 다크 테마에서도 흰 배경으로 읽히는지 확인합니다.

```python
# 표시 확인: 가상 예시를 하이라이트해 HTML 위젯 하나로 보여 준다.
import ipywidgets as widgets
from IPython.display import display
import render

source = "Please click the Save button.\nThe file will be saved  automatically. <script>alert(1)</script>"
translation = "저장 버튼을 클릭해.\n파일은 자동으로 저장됩니다. 저장 버튼"
spans_src = [(1, "click the Save button"), (2, "saved automatically")]
spans_tr = [(1, "클릭해"), (3, "저장 버튼")]

body = render.columns(render.highlight(source, spans_src), render.highlight(translation, spans_tr))
body += render.details("행 2 상세 (펼치기)", "<p>접힌 내용입니다. 줄바꿈과 &lt;태그&gt; 이스케이프 확인.</p>")
display(widgets.HTML(render.page(body)))
```
